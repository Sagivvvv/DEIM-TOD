```python
from pathlib import Path
import os, subprocess, sys, zipfile

markers = list(Path('/kaggle/input').rglob('P4B_UPLOAD.json'))
if not markers:
    archives = list(Path('/kaggle/input').rglob('p4b_kaggle_upload.zip'))
    if len(archives) != 1:
        raise RuntimeError(f'Please attach one P4-B upload dataset; archives={archives}')
    destination = Path('/kaggle/working/p4b_code')
    with zipfile.ZipFile(archives[0]) as z:
        z.extractall(destination)
    markers = list(destination.rglob('P4B_UPLOAD.json'))
if len(markers) != 1:
    raise RuntimeError(f'Expected one P4-B code dataset; found {markers}')
root = markers[0].parent
print('P4-B code:', root)
subprocess.run([sys.executable, '-B', str(root/'tools/kaggle_run_p4b.py')],
               cwd=root, env=dict(os.environ, PYTHONUNBUFFERED='1', PYTHONDONTWRITEBYTECODE='1'), check=True)
```
